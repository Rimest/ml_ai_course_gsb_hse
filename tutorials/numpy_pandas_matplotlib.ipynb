```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=3, suppress=True)
plt.rcParams.update({"figure.figsize": (7.2, 3.8), "figure.dpi": 110,
                     "axes.grid": True, "grid.alpha": 0.25,
                     "axes.spines.top": False, "axes.spines.right": False})
BLUE, RED = "#0057b8", "#c0392b"


def check(ok, hint):
    "Самопроверка задания."
    if not ok:
        raise AssertionError("✗ " + hint)
    print("✓ верно")


# ── учебный архив велопроката ──────────────────────────────────────────────
# 365 дней работы городского велопроката.
rng = np.random.default_rng(20260911)
n_days = 365

dates = pd.date_range("2025-01-01", periods=n_days)
day = np.arange(n_days)
temp = np.round(12 + 14 * np.sin(2 * np.pi * (day - 100) / 365)
                + rng.normal(0, 2.5, n_days), 1)
rain = np.round(np.clip(rng.gamma(1.3, 1.5, n_days) - 1.0, 0, None), 1)
weekend = np.asarray(dates.dayofweek >= 5)
HOLIDAYS = {"2025-01-01", "2025-01-07", "2025-02-23", "2025-03-08",
            "2025-05-01", "2025-05-09", "2025-06-12", "2025-11-04"}
holiday = np.array([d.strftime("%Y-%m-%d") in HOLIDAYS for d in dates])

rides = (550 + 48 * temp - 70 * rain + 180 * weekend - 120 * holiday
         + rng.normal(0, 90, n_days))
rides = np.clip(np.round(rides), 0, None).astype(int)

DAY_NAMES = ["понедельник", "вторник", "среда", "четверг",
             "пятница", "суббота", "воскресенье"]
bikes = pd.DataFrame({
    "Дата": dates,
    "Температура": temp,
    "Осадки": rain,
    "День недели": [DAY_NAMES[d] for d in dates.dayofweek],
    "Выходной": weekend,
    "Праздник": holiday,
    "Поездки": rides,
})
bikes["Погода"] = np.where(bikes["Осадки"] > 0.5, "дождь",
                           np.where(bikes["Температура"] > 18, "ясно", "облачно"))

stations = pd.DataFrame({
    "Велосипедов": [24, 40, 18, 32, 12, 26],
    "Свободных мест": [6, 10, 12, 8, 20, 4],
    "Район": ["Центр", "Центр", "Север", "Север", "Юг", "Юг"],
}, index=["Парк", "Вокзал", "Набережная", "Университет", "Стадион", "Рынок"])
stations.index.name = "Станция"

print("Архив велопроката:", bikes.shape[0], "дней,", bikes.shape[1], "столбцов")
bikes.head()
```
