# Часть 1. Горячие клавиши

## Главное: два режима

Почти все жалобы вида «нажимаю B, а печатается буква» объясняются одним:
у ячейки есть два состояния, и клавиши в них означают разное.

Командный режим и режим редактирования

* **Командный режим** — рамка синяя, курсора в ячейке нет. Клавиши здесь
  не печатают, а выполняют команды: `B` создаёт ячейку, `M` превращает её
  в текстовую. Попасть сюда: **Esc**.
* **Режим редактирования** — рамка зелёная, внутри мигает курсор. Клавиши
  печатают символы. Попасть сюда: **Enter** или щелчок мышью по ячейке.

Правило простое: **прежде чем нажимать команду, нажмите Esc.**

## Запуск ячеек

Работает в обоих режимах и одинаково в Jupyter и в Colab.

| Клавиши | Что делает |
|---|---|
| ★ `Shift + Enter` | выполнить ячейку и перейти к следующей |
| `Ctrl + Enter` | выполнить и остаться на месте — удобно переделывать одно и то же |
| `Alt + Enter` | выполнить и вставить пустую ячейку следом |

## Если что-то забыли

## Если всё сломалось

Порядок действий, который спасает в девяти случаях из десяти:

1. `Esc` `0` `0` — перезапустить ядро (в Colab `Ctrl + M` `.`);
2. выполнить ноутбук сверху вниз: **Runtime → Restart and run all**;
3. если ошибка осталась — читать **последнюю строку** красного блока,
   она называет причину.

Чаще всего причина в том, что ячейки выполнялись не по порядку: переменная
из ячейки ниже была создана раньше, чем та, что выше. Номера в квадратных
скобках слева показывают реальный порядок выполнения.

## Служебные команды ноутбука

Это не клавиши, а специальные команды: `%` действует на строку,
`%%` — на всю ячейку и должна стоять её первой строкой.

| Команда | Что делает |
|---|---|
| `%%time` | покажет, сколько выполнялась ячейка |
| `%timeit выражение` | замерит время, повторив выражение много раз |
| `!pip install библиотека` | установит библиотеку прямо из ноутбука |
| `%matplotlib inline` | графики рисуются под ячейкой (в Colab уже включено) |
| `%who` | список созданных переменных |
| `функция?` | полная справка по функции |
| `функция??` | справка вместе с исходным кодом |

````python
%%time
total = sum(i * i for i in range(200_000))
print("Сумма квадратов:", total)
````

# Часть 2. Команды библиотек

Всюду ниже подразумеваются стандартные сокращения:

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

В таблицах `arr` — массив numpy, `df` — таблица pandas, `ax` — оси графика.

## numpy

### Создать массив

| Команда | Что получится |
|---|---|
| `np.array([1, 2, 3])` | массив из списка |
| `np.arange(0, 10, 2)` | `0 2 4 6 8` — как `range`, но массив |
| `np.linspace(0, 1, 5)` | 5 точек от 0 до 1 включительно |
| `np.zeros(5)`, `np.ones((2, 3))` | нули и единицы заданной формы |
| `np.full((2, 2), 7)` | массив, заполненный одним числом |
| `np.eye(3)` | единичная матрица |
| `np.empty_like(arr)`, `np.zeros_like(arr)` | той же формы, что `arr` |

### Форма и перестройка

| Команда | Что делает |
|---|---|
| `arr.shape` | форма: `(строк, столбцов)` |
| `arr.ndim`, `arr.size`, `arr.dtype` | число измерений, число элементов, тип |
| `arr.reshape(3, 4)` | изменить форму |
| `arr.reshape(-1, 1)` | превратить в столбец; `-1` — «посчитай сам» |
| `arr.ravel()`, `arr.flatten()` | вытянуть в одну строку |
| `arr.T` | транспонировать: строки станут столбцами |
| `np.column_stack([a, b])` | склеить векторы как столбцы матрицы |
| `np.concatenate([a, b])`, `np.vstack`, `np.hstack` | склеить массивы |
| `arr.astype(float)` | сменить тип элементов |

### Отобрать элементы

| Команда | Что делает |
|---|---|
| `arr[0]`, `arr[-1]` | первый и последний |
| `arr[1:4]` | срез: элементы 1, 2, 3 |
| `arr[:, 2]` | весь третий **столбец** матрицы |
| `arr[1, :]` | вся вторая **строка** |
| `arr[arr > 5]` | все элементы, удовлетворяющие условию |
| `(arr > 5).sum()` | сколько их — `True` считается за единицу |
| `(arr > 5).any()`, `.all()` | есть ли хоть один; все ли |
| `np.where(arr > 5, 1, 0)` | заменить по условию: да → 1, нет → 0 |
| `np.argmax(arr)`, `np.argmin(arr)` | **номер** максимума и минимума |
| `np.argsort(arr)` | номера элементов в порядке возрастания |

### Посчитать

| Команда | Что делает |
|---|---|
| `arr.sum()`, `.mean()`, `.std()`, `.min()`, `.max()` | по всему массиву |
| `arr.mean(axis=0)` | по столбцам: одно число на столбец |
| `arr.mean(axis=1)` | по строкам: одно число на строку |
| `np.median(arr)`, `np.percentile(arr, 95)` | медиана и квантиль |
| `np.cumsum(arr)` | накопленная сумма |
| `np.round(arr, 2)`, `np.clip(arr, 0, 10)` | округлить; обрезать по краям |
| `np.abs`, `np.sqrt`, `np.exp`, `np.log` | поэлементные функции |
| `np.unique(arr)` | уникальные значения |
| `np.unique(arr, return_counts=True)` | уникальные и сколько раз встретились |
| `np.isnan(arr)`, `np.nanmean(arr)` | найти пропуски; среднее в обход пропусков |

### Линейная алгебра и случайность

| Команда | Что делает |
|---|---|
| `a * b` | поэлементное умножение |
| `a @ b` | скалярное произведение векторов или умножение матриц |
| `np.linalg.norm(a)` | длина вектора |
| `np.linalg.inv(A)`, `np.linalg.solve(A, b)` | обратная матрица; решение системы |
| `rng = np.random.default_rng(42)` | генератор с фиксированным зерном |
| `rng.normal(0, 1, size=100)` | нормальные случайные числа |
| `rng.integers(0, 10, size=5)`, `rng.random(5)` | целые; равномерные от 0 до 1 |
| `rng.choice(arr, 3, replace=False)` | случайная выборка без повторов |
| `rng.permutation(arr)` | перемешать |
| `np.allclose(a, b)` | равны ли массивы с точностью до округления |

## pandas

### Прочитать и сохранить

| Команда | Что делает |
|---|---|
| `pd.read_csv("f.csv")` | прочитать таблицу из файла |
| `pd.read_csv("f.csv", sep=";", decimal=",")` | русский формат: точка с запятой и запятая |
| `pd.read_excel("f.xlsx", sheet_name="Лист1")` | прочитать из Excel |
| `pd.DataFrame({"A": [1, 2], "B": [3, 4]})` | собрать таблицу из словаря |
| `df.to_csv("f.csv", index=False)` | сохранить без столбца индекса |

### Осмотреть

| Команда | Что делает |
|---|---|
| `df.shape` | сколько строк и столбцов |
| `df.head(5)`, `df.tail(5)`, `df.sample(5)` | первые, последние, случайные строки |
| `df.columns`, `df.dtypes` | имена столбцов и их типы |
| `df.info()` | размер, типы и число заполненных значений |
| `df.describe()` | среднее, разброс, минимум, максимум, квартили |
| `df["A"].value_counts()` | сколько раз встречается каждое значение |
| `df["A"].nunique()`, `df["A"].unique()` | сколько разных значений; какие |
| `df.corr(numeric_only=True)` | матрица корреляций |

### Выбрать столбцы и строки

| Команда | Что делает |
|---|---|
| `df["A"]` | один столбец |
| `df[["A", "B"]]` | несколько столбцов — список внутри скобок |
| `df.loc[0, "A"]` | значение по **имени** строки и столбца |
| `df.iloc[0, 1]` | значение по **номеру** строки и столбца |
| `df.loc[маска, "A"]` | столбец только у отобранных строк |
| `df.iloc[:5]`, `df.iloc[:, 1:3]` | срез по номерам строк или столбцов |

### Отфильтровать

| Команда | Что делает |
|---|---|
| `df[df["A"] > 5]` | строки, где выполняется условие |
| `df[(df["A"] > 5) & (df["B"] < 2)]` | два условия, каждое в своих скобках |
| `df[(df["A"] > 5) \| (df["B"] < 2)]` | «или» — вертикальная черта |
| `df[~df["A"].isna()]` | отрицание условия — тильда |
| `df[df["A"].isin([1, 2, 3])]` | значение входит в список |
| `df[df["A"].between(1, 5)]` | значение в диапазоне |
| `df[df["Город"].str.contains("Моск")]` | текстовый столбец содержит подстроку |
| `df.query("A > 5 and B < 2")` | то же условие строкой |

### Изменить и добавить

| Команда | Что делает |
|---|---|
| `df["Новый"] = df["A"] * 2` | новый столбец |
| `df["Флаг"] = df["A"] > 5` | логический столбец |
| `df["Класс"] = np.where(df["A"] > 5, "высокий", "низкий")` | по условию |
| `df["A"].apply(функция)` | применить свою функцию к каждому значению |
| `df["A"].map({1: "да", 0: "нет"})` | заменить значения по словарю |
| `df.rename(columns={"A": "Новое имя"})` | переименовать столбцы |
| `df.drop(columns=["A", "B"])` | удалить столбцы |
| `df.astype({"A": float})` | сменить тип столбца |
| `pd.cut(df["A"], bins=[0, 10, 20, 100])` | разбить на интервалы |

### Сгруппировать и свести

| Команда | Что делает |
|---|---|
| `df.groupby("K")["V"].mean()` | среднее внутри каждой группы |
| `df.groupby("K")["V"].agg(["mean", "count", "std"])` | несколько показателей сразу |
| `df.groupby("K").agg({"V": "sum", "U": "mean"})` | своя функция для каждого столбца |
| `df.groupby(["K1", "K2"])["V"].mean()` | группировка по двум ключам |
| `df.groupby("K").size()` | размер каждой группы |
| `df.groupby("K")["V"].transform("mean")` | среднее группы, размноженное по строкам |
| `pd.pivot_table(df, index="A", columns="B", values="V", aggfunc="mean")` | сводная таблица |
| `df.sort_values("A", ascending=False)` | сортировка по столбцу |
| `df.nlargest(5, "A")`, `df.nsmallest(5, "A")` | пять наибольших или наименьших |

### Пропуски и дубликаты

| Команда | Что делает |
|---|---|
| `df.isna().sum()` | сколько пропусков в каждом столбце |
| `df.dropna()`, `df.dropna(subset=["A"])` | выбросить строки с пропусками |
| `df.fillna(0)`, `df.fillna(df.median())` | заполнить нулём или медианой |
| `df["A"].ffill()` | заполнить предыдущим значением |
| `df.duplicated().sum()`, `df.drop_duplicates()` | найти и убрать дубликаты |

### Даты и объединение таблиц

| Команда | Что делает |
|---|---|
| `pd.to_datetime(df["Дата"])` | превратить текст в дату |
| `df["Дата"].dt.year`, `.dt.month`, `.dt.dayofweek` | части даты |
| `pd.date_range("2025-01-01", periods=365)` | последовательность дат |
| `df.set_index("Дата").resample("ME").mean()` | помесячные средние; в pandas до 2.2 писали `"M"` |
| `pd.merge(a, b, on="id", how="left")` | соединить таблицы по ключу |
| `pd.concat([a, b])`, `pd.concat([a, b], axis=1)` | склеить по строкам или столбцам |

### Выйти в numpy

| Команда | Что делает |
|---|---|
| `df.to_numpy()`, `df[["A", "B"]].to_numpy()` | матрица признаков для модели |
| `df["A"].to_numpy()` | один столбец как массив |
| `pd.set_option("display.max_columns", 50)` | показывать больше столбцов |

## matplotlib

### Каркас

| Команда | Что делает |
|---|---|
| `fig, ax = plt.subplots(figsize=(8, 4))` | одна система координат — так и надо начинать |
| `fig, axes = plt.subplots(1, 3, figsize=(12, 3))` | три графика в ряд, обращение `axes[0]` |
| `fig, axes = plt.subplots(2, 2)` | сетка 2 × 2, обращение `axes[0, 1]` |
| `plt.tight_layout()` | развести подписи, чтобы не наезжали |
| `plt.show()` | показать; ставится последней строкой |
| `fig.savefig("plot.png", dpi=150, bbox_inches="tight")` | сохранить в файл |

### Типы графиков

| Команда | Когда нужен |
|---|---|
| `ax.plot(x, y)` | величина во времени, кривая |
| `ax.scatter(x, y, s=10, alpha=.5)` | связь двух величин, облако точек |
| `ax.hist(x, bins=30)` | как распределена одна величина |
| `ax.bar(x, h)`, `ax.barh(y, w)` | сравнение категорий |
| `ax.boxplot([a, b])` | сравнение распределений: медиана и квартили |
| `ax.axhline(y, ls="--")`, `ax.axvline(x)` | горизонтальная и вертикальная линия — порог, среднее |
| `ax.fill_between(x, lo, hi, alpha=.2)` | закрасить коридор между кривыми |
| `ax.contour(X, Y, Z)` | линии уровня поверхности |

### Оформление

| Команда | Что делает |
|---|---|
| `ax.set_xlabel("подпись")`, `ax.set_ylabel(...)` | подписи осей — **обязательны** |
| `ax.set_title("заголовок")` | заголовок графика |
| `ax.legend(frameon=False)` | легенда; работает, если задан `label=` у кривых |
| `ax.set_xlim(0, 10)`, `ax.set_ylim(...)` | границы осей |
| `ax.grid(alpha=.3)` | сетка |
| `ax.set_xticks(...)`, `ax.tick_params(rotation=45)` | свои отметки; повернуть подписи |
| `ax.set_yscale("log")` | логарифмическая шкала |
| `color=`, `lw=`, `ls="--"`, `marker="o"`, `alpha=` | цвет, толщина, пунктир, точки, прозрачность |
| `plt.rcParams.update({...})` | оформление сразу для всех графиков ноутбука |

### Быстрый график прямо из pandas

| Команда | Что делает |
|---|---|
| `df["A"].hist(bins=30)` | гистограмма столбца |
| `df.plot(x="Дата", y="Осадки", ax=ax)` | линия из таблицы |
| `df.plot.scatter(x="A", y="B", ax=ax)` | точечный график |
| `df.groupby("K")["V"].mean().plot.bar(ax=ax)` | столбцы по группам |
